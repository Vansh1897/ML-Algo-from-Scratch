```python
import pandas as pd
import numpy as np
```

### Custom Multi Linear Regression Function

```python
class multiLinearRegression:
    def __init__(self):
        self.w = None
        self.coef_ = None
        self.intercept = None
    
    def fit(self,x, y):
        x = np.insert(x,0,1, axis=1)
        self.w = np.ones(x.shape[1])
        w = np.linalg.inv(x.T.dot(x)).dot(x.T).dot(y)
        self.coef_ = w[1:]
        self.intercept_ = w[0]

    def predict(self,x_test):
        return np.array(self.intercept_ + x_test.dot(self.coef_))
    
```

```python
from sklearn.datasets import load_diabetes
df = load_diabetes(scaled=True,as_frame=True)
df = pd.DataFrame(pd.concat([df.data, df.target.rename('target')], axis = 1))
df.head()
```

```python
from sklearn.model_selection import train_test_split
x = df.drop(columns = 'target')
y = df['target']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=.2, random_state=32)
```

#### Sklearn Linear Regression

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
lr.intercept_,lr.coef_
```

```python
lr.predict(x_test)
```

#### Custom Linear Regression

```python
lr1 = multiLinearRegression()
lr1.fit(x_train,y_train)
lr1.intercept_, lr1.coef_
```

```python
lr1.predict(x_test)
```
