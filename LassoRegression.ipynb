### Custom Lasso Regression Function 
##### Still figuring out how this works in sklearn (this is just the general implementation of Lasso Regression based on the basic understanding/functionality)

```python
class CustomLasso:
    def __init__(self, alpha):
        self.w = None
        self.intercept_ = None
        self.coef_ = None
        self.lr = 0.01
        self.alpha = alpha

    def fit(self, x, y, epochs = 1000):
        dw = 0
        da = 0
        x = np.insert(x,0,1, axis =1)
        self.w = np.ones(x.shape[1])
        i = np.ones(x.shape[1])
        i[0] = 0
        for _ in range(epochs):
            y_pred = np.dot(x, self.w)
            dw = (1/x.shape[0])*(x.T.dot(y_pred - y))
            dw += self.alpha*np.sign(self.w)*i

            self.w = self.w- self.lr*dw

        self.intercept_ = self.w[0]
        self.coef_ = self.w[1:]

    def predict(self,x_test):
        return np.array(np.dot(x_test, self.coef_) + self.intercept_).flatten()


    
```

```python
from sklearn.datasets import load_diabetes
df = load_diabetes(scaled=True,as_frame=True)
df = pd.DataFrame(pd.concat([df.data, df.target.rename('target')], axis = 1))


from sklearn.model_selection import train_test_split
x = df.drop(columns = 'target')
y = df['target']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=.2, random_state=32)
```

#### Sklearn Lasso Regression

```python
from sklearn.linear_model import Lasso
model = Lasso(2, tol = 0)
model.fit(x_train, y_train)
model.intercept_, model.coef_
```

```python
model.predict(x_test)
```

#### Custom Lasso Regression

```python
model1 = CustomLasso(2)
model1.fit(x_train, y_train)
model1.intercept_, model1.coef_
```

Here the intercept is almost same as sklearn lasso whereas the coef values are different but it performs the same task as irrelavant feature's weights minimizes/reduces towards 0 whereas the important features have more weights values.

```python
model1.predict(x_test)
```
