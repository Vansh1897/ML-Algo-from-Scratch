```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeRegressor
```

### Custom Gradient Boosting Classifier Function

```python
class GradientBoostingClassifier:
    def __init__(self, n_estimators = 100, max_depth = 8, learning_rate = 0.1):
        self.max_depth = max_depth
        self.n_estimators = n_estimators
        self.lr = learning_rate
        self.intial_pred = None
        self.trees = []

    def sigmoid(self, pred):
        return 1/(1 + np.exp(-pred))

    def fit (self, x, y):
        p = np.mean(y)
        self.initialpred = np.log(p/(1-p))
        y_pred = np.full(y.shape[0], self.initialpred)

        for _ in range(self.n_estimators):
            prob = self.sigmoid(y_pred)
            res = y - prob

            tree = DecisionTreeRegressor(max_depth = self.max_depth)
            tree.fit(x,res)
            pred = tree.predict(x)
            ypred = y_pred + (self.lr * pred)
            self.trees.append(tree)

    def predict_proba(self, x_test):
        probablities = np.full(x_test.shape[0], self.initialpred)
        for tree in self.trees:
            probablities = probablities + (self.lr * tree.predict(x_test))

        return self.sigmoid(probablities)

    def predict(self, x_test):
        prediction = self.predict_proba(x_test)
        return np.where(prediction>=0.5, 1, 0)
```

```python
from sklearn.datasets import load_breast_cancer
df = load_breast_cancer(as_frame=True)
df = pd.DataFrame(pd.concat([df.data, df.target.rename('target')], axis = 1))


from sklearn.model_selection import train_test_split
x = df.drop(columns = 'target')
y = df['target']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=.2, random_state=32)
```

#### Sklearn Gradient Boosting Classifier

```python
from sklearn.ensemble import GradientBoostingClassifier
model = GradientBoostingClassifier(n_estimators = 100, max_depth = 8)
model.fit(x_train, y_train)
model.predict(x_test)
```
