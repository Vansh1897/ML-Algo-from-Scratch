### Custom Naive Bayes Function

```python
class naive_bayes:    
    def fit(self,x,y):
        self.n_rows, self.n_cols = x.shape
        self.classes = y.unique()
        self.n_classes = len(self.classes)

        self.mean = np.zeros((self.n_classes, self.n_cols), dtype=np.float64)
        self.var = np.zeros((self.n_classes, self.n_cols), dtype=np.float64)
        self.priors = np.zeros(self.n_classes, dtype = np.float64)

        for i, c in enumerate(self.classes):
            X_c = x[y==c]
            self.mean[i,:] = X_c.mean(axis = 0) 
            self.var[i,:] = X_c.var(axis = 0)
            self.priors[i] = len(X_c)/ self.n_rows
    
    def predict(self, x_test):
        prediction = [self.get_prediction(x_test.iloc[i]) for i in range(len(x_test))]
        return np.array(prediction)

    def get_prediction(self, x):
        posteriors = []

        for i, c in enumerate(self.classes):
            prior = np.log(self.priors[i])
            posterior  = np.sum(np.log(self.pdf(i, x)))
            posterior = posterior+prior
            posteriors.append(posterior)
        
        return self.classes[np.argmax(posteriors)]

    def pdf(self,i, x):
        mean, var = self.mean[i], self.var[i]
        x = x.astype(np.float64)
        num = np.exp(-((x-mean)**2) / (2*var))     
        den = np.sqrt(2*np.pi*var)    
        return num/den
               

        
            
```

```python
x = df.drop(columns = 'PlayTennis')
y = df['PlayTennis']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, random_state = 2, test_size = 0.3)

x_train, x_test, y_train, y_test = x_train.astype(int), x_test.astype(int), y_train.astype(int), y_test.astype(int)
```

#### Sklearn Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB
nb = GaussianNB()
nb.fit(x_train, y_train)
nb.predict(x_test)
```
