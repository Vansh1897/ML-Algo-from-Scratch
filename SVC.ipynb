```python
import pandas as pd
import numpy as np
```

### Custom SVC Function

```python
class SVM:
    def __init__(self,n_iter = 1000, lr = 0.001, lambda1 = 0.01):
        self.n_iter = n_iter
        self.lr = lr
        self.lambda1 = lambda1
        self.w = None
        self.b = None

    def fit(self, X, y):
        X = np.array(X, dtype=np.float64)
        y = np.array(y, dtype=np.float64)

        n_samples, n_features = X.shape
        y_ = np.where(y<=0, -1, 1)
        self.w = np.zeros(n_features)
        self.b = 0

        for _ in range(self.n_iter):
            for i, x in enumerate(X):
                condition = y_[i] * (np.dot(x,self.w) - self.b) >= 1
                if condition:
                    self.w = self.w - self.lr * (2 * self.lambda1 * self.w)
                else:
                    self.w = self.w - self.lr * ((2 * self.lambda1 * self.w) - y_[i] * x)
                    self.b = self.b - self.lr * y_[i]

    def predict(self, x_test):
        prediction = np.dot(x_test, self.w) - self.b
        prediction = np.sign(prediction)
        return np.where(prediction == -1, 0, 1)
```

```python
from sklearn.datasets import load_breast_cancer
df = load_breast_cancer(as_frame=True)
df = pd.DataFrame(pd.concat([df.data, df.target.rename('target')], axis = 1))


from sklearn.model_selection import train_test_split
x = df.drop(columns = 'target')
y = df['target']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=.2, random_state=32)

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

#### Sklearn SVC

```python
from sklearn.svm import SVC
model = SVC(max_iter = 1000)
model.fit(x_train, y_train)
model.predict(x_test)
```

#### Custom SVC

```python
model1 = SVM()
model1.fit(x_train, y_train)
model1.predict(x_test)
```
