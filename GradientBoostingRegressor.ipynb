```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeRegressor
```

### Custom Gradient Boosting Function

```python
class GradientBoosting:
    def __init__(self,n_estimators = 100, max_depth = 3):
        self.n_estimators = n_estimators
        self.max_depth = max_depth
        self.learning_rate = 0.1
        self.trees = []
        self.initialpred = None

    def fit(self, x, y):
        self.initialpred = np.mean(y)
        y_pred = np.full(y.shape[0], self.initialpred)
        
        for _ in range(self.n_estimators):
            res = y - y_pred

            tree = DecisionTreeRegressor(max_depth = self.max_depth)
            tree.fit(x, res)
            pred = tree.predict(x)

            y_pred = y_pred + self.learning_rate * pred

            self.trees.append(tree)

    def predict(self, x_test): 
        y_pred = np.full(x_test.shape[0], self.initialpred)

        for tree in self.trees:
            y_pred += self.learning_rate * tree.predict(x_test)

        return np.array(y_pred)

        
```

```python
from sklearn.datasets import load_diabetes
df = load_diabetes(scaled=True,as_frame=True)
df = pd.DataFrame(pd.concat([df.data, df.target.rename('target')], axis = 1))
from sklearn.model_selection import train_test_split
x = df.drop(columns = 'target')
y = df['target']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=.2, random_state=32)
```

#### Sklearn Gradient Boosting Regressor

```python
from sklearn.ensemble import GradientBoostingRegressor
model = GradientBoostingRegressor(n_estimators = 100, max_depth = 3)
model.fit(x_train, y_train)
model.predict(x_test)
```

#### Custom Gradient Boosting Regressor

```python
model1 = GradientBoosting()
model1.fit(x_train, y_train)
model1.predict(x_test)
```
