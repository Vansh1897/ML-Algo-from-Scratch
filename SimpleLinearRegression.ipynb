```python
import pandas as pd
import numpy as np
```

### Custom Linear Regression Function (for 1 dependant and 1 independant feature)

```python
class LinearRegressionCustom:
    def __init__(self):
        self.coef_ = None
        self.intercept_= None

    def fit(self,x,y):
        num = 0
        den = 0
        x = x.values.flatten()
        y = y.values.flatten()

        for i in range(x.shape[0]):
            num = num + (x[i]-x.mean())*(y[i]-y.mean())
            den = den + (x[i]-x.mean())**2
        self.coef_ = num/den
        self.intercept_= y.mean()-self.coef_*x.mean()

    def predict(self, x_test):
        return np.array(self.coef_*x_test + self.intercept_).flatten()
```

```python
from sklearn.datasets import load_diabetes
```

```python
df = load_diabetes(as_frame=True, scaled=False)
```

```python

df = pd.DataFrame(pd.concat([df.data,df.target.rename('Target')], axis=1))
```

```python
x = df[['bmi']]
y = df.Target
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=32)
```

#### Sklearn Linear Regression

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred = lr.predict(x_test)
```

```python
lr.coef_, lr.intercept_
```

```python
lr.predict(x_test)
```

#### Custom Linear Regression 

```python
lr1 = LinearRegressionCustom()
lr1.fit(x_train, y_train)
```

```python
lr1.coef_, lr1.intercept_
```

```python
lr1.predict(x_test)
```
