```python
import pandas as pd 
import numpy as np
from collections import Counter
```

### Custom K Nearest Neighbour(classifier) Function

```python
class KNN:
    def __init__(self,k = 3):
        self.k = k
    
    def fit(self,x,y):
        self.x_train = x
        self.y_train = y

    def predict(self, x_test):
        predictions = [self.get_prediction(x_test.iloc[i]) for i in range(len(x_test))] 
        return np.array(predictions)
    
    def euclidian_dist(self,x,y):
        return np.sqrt(np.sum((y-x)**2))
    
    def get_prediction(self,input):
        distances = [self.euclidian_dist(input, self.x_train.iloc[i]) for i in range(len(self.x_train))]
        top_k = np.argsort(distances)[:self.k]
        top_k_labels = [self.y_train.iloc[i] for i in top_k]
        prediction = Counter(top_k_labels).most_common(1)[0][0]
        
        return prediction


        
            
```

```python

url = 'https://raw.githubusercontent.com/hayesall/bn-rule-extraction/main/toy_decision.csv'
df = pd.read_csv(url)

```

```python
x = df.drop(columns = 'PlayTennis')
y = df['PlayTennis']
```

```python
x = pd.get_dummies(x)
y = y.map({'yes' : 1, 'no': 0})
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, random_state = 2, test_size = 0.3)
x_train, x_test, y_train, y_test = x_train.astype(int), x_test.astype(int), y_train.astype(int), y_test.astype(int)
```

#### Sklearn KNN Classifier

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors = 3)
model.fit(x_train, y_train)
model.predict(x_test)
```

#### Custom KNN

```python
model1 = KNN(k=3)
model1.fit(x_train, y_train)
model1.predict(x_test)
```
