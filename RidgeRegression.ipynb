```python
import pandas as pd
import numpy as np
```

### Custom Ridge Regression Function

```python
class CustomRidge:
    def __init__(self,alpha = 0):
        self.w = None
        self.coef_ = None
        self.intercept_ = None
        self.alpha = alpha

    def fit(self,x,y):
        x = np.insert(x,0,1,axis = 1)
        self.w = np.ones(x.shape[1])
        i = np.identity(x.shape[1])
        i[0][0] = 0
        
        self.w = x.T.dot(y).dot(np.linalg.inv(x.T.dot(x)+ (self.alpha*i)))
        self.intercept_ = self.w[0]
        self.coef_ = self.w[1:]

    def predict(self,x_test):
        return np.array(x_test.dot(self.coef_) + self.intercept_).flatten()
```

```python
from sklearn.datasets import load_diabetes
df = load_diabetes(scaled=True,as_frame=True)
df = pd.DataFrame(pd.concat([df.data, df.target.rename('target')], axis = 1))


from sklearn.model_selection import train_test_split
x = df.drop(columns = 'target')
y = df['target']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=.2, random_state=32)
```

#### Sklearn Ridge Regression

```python
from sklearn.linear_model import Ridge
model = Ridge(2)
model.fit(x_train, y_train)
model.intercept_, model.coef_
```

```python
model.predict(x_test)
```

#### Custom Ridge Regression

```python
model1 = CustomRidge(2)
model1.fit(x_train, y_train)
model1.intercept_, model1.coef_
```

```python
model1.predict(x_test)
```
